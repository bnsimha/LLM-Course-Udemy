```python
import re

text = [ 'All that we are is the result of what we have thought',
         'To be or not to be that is the question',
         'Be yourself everyone else is already taken' ]

all_words = re.split(r'\s',' '.join(text).lower())
vocab = sorted(set(all_words))
```

```python
#encoder

text2idx = { text : id for id,text in enumerate(vocab)}
```

```python
#decoder

idx2text = { id : text for id,text in enumerate(vocab)}
```

```python
#wrapping up in funnctions

def encode(text):
  words = re.split(r'\s',text.lower())
  ids = [ text2idx[word] for word in words ]
  return ids

def decode(ids):
  words = [ idx2text[id] for id in ids ]
  text = ' '.join(words)
  return text
```

```python
newtext = 'we already are the result of what ' \
          'everyone else already thought'
newtext_tokenIDs = encode(newtext)
decoded_text = decode(newtext_tokenIDs)

print('Original text:')
print(f'\t{newtext}')

print(f'\nToken IDs:')
print(f'\t{newtext_tokenIDs}')

print(f'\nDecoded text:')
print(f'\t{decoded_text}')
```

```python
target_word = 'to'
target_id = text2idx[target_word]

full_text = ' '.join(text).lower()
words = re.split(r'\s', full_text)
idx = np.array([ text2idx[word] for word in words ])
index = np.where(idx == target_id)
final = [ words[max(0, t-1) : min(len(words), t+2)] for t in index[0] ]
string = [ ' '.join(f) for f in final ]
string
```
